## 01. Perceptrón Simple y Red de Una Capa (Clasificación Binaria)
En este notebook se implementan desde cero el Perceptrón de Rosenblatt y una Red Monocapa con función de activación Sigmoide.


```python
import numpy as np
import matplotlib.pyplot as plt

class Perceptron:
    def __init__(self, lr=0.1, n_iter=100):
        self.lr, self.n_iter = lr, n_iter
        self.w, self.b = None, 0.0

    def fit(self, X, y):
        self.w = np.zeros(X.shape[1])
        for epoch in range(self.n_iter):
            errors = 0
            for xi, target in zip(X, y):
                y_p = 1 if (np.dot(xi, self.w) + self.b) >= 0 else 0
                update = self.lr * (target - y_p)
                self.w += update * xi
                self.b += update
                errors += int(update != 0)
            if errors == 0:
                print(f"[Perceptrón] Convergencia en la época {epoch+1}")
                break
        return self

    def predict(self, X):
        return np.where((np.dot(X, self.w) + self.b) >= 0, 1, 0)

# Verificación con compuerta lógica AND
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0, 0, 0, 1])
p = Perceptron(lr=0.1, n_iter=20).fit(X, y)
print("Predicciones AND:", p.predict(X))
```

```python
class SingleLayerNN:
    def __init__(self, lr=0.05, n_epochs=500):
        self.lr, self.n_epochs = lr, n_epochs
        self.loss_history = []

    def _sig(self, z):
        return 1.0 / (1.0 + np.exp(-np.clip(z, -250, 250)))

    def fit(self, X, y):
        m, n = X.shape
        self.w = np.random.randn(n) * 0.01
        self.b = 0.0
        for _ in range(self.n_epochs):
            a = self._sig(np.dot(X, self.w) + self.b)
            loss = -(1.0/m) * np.sum(y*np.log(a+1e-15) + (1-y)*np.log(1-a+1e-15))
            self.loss_history.append(loss)
            dz = a - y
            self.w -= self.lr * (1.0/m) * np.dot(X.T, dz)
            self.b -= self.lr * (1.0/m) * np.sum(dz)
        return self

    def predict(self, X):
        return (self._sig(np.dot(X, self.w) + self.b) >= 0.5).astype(int)

snn = SingleLayerNN().fit(X, y)
print("Predicciones Red Monocapa:", snn.predict(X))
```
